##Coin change - Greedy technique

```python
#Coin change - Greedy technique
################################################################################
SYSTEM = [11, 5 ,  1  ]
################################################################
def coin_change(AMOUNT,SYSTEM):
#....
  SOLUTION = [0]*len(SYSTEM)
  accumulated_value = 0

  for i,value in enumerate(SYSTEM):
    coins = (AMOUNT-accumulated_value)//value
    SOLUTION[i] = coins
    accumulated_value = accumulated_value + coins*value

    if AMOUNT == accumulated_value:
      return SOLUTION

  print("It is not possible to find a solution")

coin_change(15,SYSTEM)


################################################################
```

##N Queens - Backtracking

```python
#N Queens - Backtracking()
################################################################################

#Checks that there are no threats between queens in the partial solution
################################################################
def is_promising(SOLUTION,stage):
################################################################
  #print(SOLUTION)
  #If the solution has two equal values it is not valid => Two queens in the same row
  for i in range(stage+1):
    #print("The value " + str(SOLUTION[i]) + " appears " +  str(SOLUTION.count(SOLUTION[i])) + " times")
    if SOLUTION.count(SOLUTION[i]) > 1:
      return False

    #Check the diagonals
    for j in range(i+1, stage +1 ):
      #print("Checking diagonal of " + str(i) + " and " + str(j))
      if abs(i-j) == abs(SOLUTION[i]-SOLUTION[j]) : return False
  return True

#Translates the solution to the board
################################################################
def print_board(S):
################################################################
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")



#Main N-Queens process
################################################################
def n_queens(N, solution=[],stage=0):
################################################################
### ....
  if len(solution) == 0:         # [0,0,0...]
    solution = [0 for i in range(N) ]

  for i in range(1, N+1):
    solution[stage] = i
    if is_promising(solution, stage):
      if stage == N-1:
        print(solution)
      else:
        n_queens(N, solution, stage+1)
    else:
      None

  solution[stage] = 0

n_queens(8,solution=[],stage=0)
```

##River trip - Dynamic programming

```python
#River trip - Dynamic programming
################################################################################

FARES = [
[0,5,4,3,999,999,999],
[999,0,999,2,3,999,11],
[999,999, 0,1,999,4,10],
[999,999,999, 0,5,6,9],
[999,999, 999,999,0,999,4],
[999,999, 999,999,999,0,3],
[999,999,999,999,999,999,0]
]

#999 can be replaced by float("inf")


#Computation of the PRICES and ROUTES matrices
################################################################
def Prices(FARES):
################################################################
  #Total number of nodes
  N = len(FARES[0])

  #Initialization of the price table
  PRICES = [ [9999]*N for i in [9999]*N]
  ROUTE = [ [""]*N for i in [""]*N]

  for i in range(0,N-1):
    ROUTE[i][i] = i             #To go from i to i you "pass through i"
    PRICES[i][i] = 0            #To go from i to i you pay 0
    for j in range(i+1, N):
      MIN = FARES[i][j]
      ROUTE[i][j] = i

      for k in range(i, j):
        if PRICES[i][k] + FARES[k][j] < MIN:
            MIN = min(MIN, PRICES[i][k] + FARES[k][j] )
            ROUTE[i][j] = k          #Records that to go from i to j you must pass through k
        PRICES[i][j] = MIN

  return PRICES,ROUTE
################################################################

PRICES,ROUTE = Prices(FARES)
#print(PRICES[0][6])

print("PRICES")
for i in range(len(FARES)):
  print(PRICES[i])

print("\nROUTE")
for i in range(len(FARES)):
  print(ROUTE[i])

#Determine the route using recursion
def compute_route(ROUTE, source, target):
  if source == target:
    #print("Go to :" + str(source))
    return ""
  else:
    return str(compute_route( ROUTE, source, ROUTE[source][target])) +  \
                ',' + \
                str(ROUTE[source][target] \
              )

print("\nThe route is:")
compute_route(ROUTE, 0,6)
```

##Finding the closest points

```python
import random
import time

#LIST_1D = [random.randrange(1,10000) for x in range(1000)]
LIST_1D = random.sample(range(1, 10000), 1000)   # using sample no repeated points are generated
LIST_2D = [(random.randrange(1,10000),random.randrange(1,10000)) for x in range(1000)]
LIST_3D = [(random.randrange(1,10000), random.randrange(1,10000), random.randrange(1,10000)) for x in range(1000)]
```

```python
def brute_force_1d(points):
    points = sorted(points)
    n = len(points)
    min_distance = float('inf')
    closest_points = None

    for i in range(n - 1):
        for j in range(i + 1, n):
            current_distance = abs(points[i] - points[j])
            if current_distance < min_distance:
                min_distance = current_distance
                closest_points = (points[i], points[j])

    return closest_points

result = brute_force_1d(LIST_1D)
print("Closest points:", result)
```

```python
def brute_force_2d(points):
    n = len(points)
    min_distance = float('inf')
    closest_points = None

    for i in range(n - 1):
        for j in range(i + 1, n):
            current_distance = ((points[i][0] - points[j][0])**2 + (points[i][1] - points[j][1])**2)**0.5
            if current_distance < min_distance:
                min_distance = current_distance
                closest_points = (points[i], points[j])

    return closest_points

result = brute_force_2d(LIST_2D)
print("Closest points:", result)
```

```python
def brute_force_3d(points):
    n = len(points)
    min_distance = float('inf')
    closest_points = None

    for i in range(n - 1):
        for j in range(i + 1, n):
            current_distance = ((points[i][0] - points[j][0])**2 + (points[i][1] - points[j][1])**2 + (points[i][2] - points[j][2])**2)**0.5
            if current_distance < min_distance:
                min_distance = current_distance
                closest_points = (points[i], points[j])

    return closest_points

start_time = time.time()
result = brute_force_3d(LIST_3D)
end_time = time.time()
print("Time taken: ", end_time - start_time, "seconds")
print("Closest points:", result)
```

```python
def divide_and_conquer_1d(points):
    points = sorted(points)
    n = len(points)
    if n <= 2:
        return points

    mid = n // 2
    left = divide_and_conquer_1d(points[:mid])
    right = divide_and_conquer_1d(points[mid:])

    d_left = abs(left[1] - left[0]) if len(left) == 2 else float('inf')
    d_right = abs(right[1] - right[0]) if len(right) == 2 else float('inf')
    d_cross = abs(points[mid-1] - points[mid])

    if d_left <= min(d_right, d_cross):
        return left
    elif d_right <= min(d_left, d_cross):
        return right
    else:
        return points[mid-1:mid+1]

result = divide_and_conquer_1d(LIST_1D)
print("Closest points:", result)
```

```python
def divide_and_conquer_2d(points):
    points = sorted(points, key=lambda x: x[0])
    n = len(points)
    if n <= 2:
        return points

    mid = n // 2
    left = divide_and_conquer_2d(points[:mid])
    right = divide_and_conquer_2d(points[mid:])

    d_left = ((left[0][0] - left[1][0])**2 + (left[0][1] - left[1][1])**2)**0.5 if len(left) == 2 else float('inf')
    d_right = ((right[0][0] - right[1][0])**2 + (right[0][1] - right[1][1])**2)**0.5 if len(right) == 2 else float('inf')
    d_cross = ((points[mid-1][0] - points[mid][0])**2 + (points[mid-1][1] - points[mid][1])**2)**0.5

    if d_left <= min(d_right, d_cross):
        return left
    elif d_right <= min(d_left, d_cross):
        return right
    else:
        return points[mid-1:mid+1]

result = divide_and_conquer_2d(LIST_2D)
print("Closest points:", result)
```

```python
def divide_and_conquer_3d(points):
    points = sorted(points, key=lambda x: x[0])
    n = len(points)
    if n <= 2:
        return points

    mid = n // 2
    left = divide_and_conquer_3d(points[:mid])
    right = divide_and_conquer_3d(points[mid:])

    d_left = ((left[0][0] - left[1][0])**2 + (left[0][1] - left[1][1])**2 + (left[0][2] - left[1][2])**2)**0.5 if len(left) == 2 else float('inf')
    d_right = ((right[0][0] - right[1][0])**2 + (right[0][1] - right[1][1])**2 + (right[0][2] - right[1][2])**2)**0.5 if len(right) == 2 else float('inf')
    d_cross = ((points[mid-1][0] - points[mid][0])**2 + (points[mid-1][1] - points[mid][1])**2 + (points[mid-1][2] - points[mid][2])**2)**0.5

    if d_left <= min(d_right, d_cross):
        return left
    elif d_right <= min(d_left, d_cross):
        return right
    else:
        return points[mid-1:mid+1]

start_time = time.time()
result = divide_and_conquer_3d(LIST_3D)
end_time = time.time()
print("Time taken: ", end_time - start_time, "seconds")
print("Closest points:", result)
```

It can be seen that the computational cost of the Brute Force algorithm is $O(n^2)$, since it has 2 nested loops.

If we focus on the Divide and Conquer algorithm, the computational complexity is $O(n \log n)$. Note that in the previous case the points are sorted before being compared in order to improve performance. Logarithmic computational complexity is characteristic of recursion.
