>1. Dubbing Sessions <br>

The dubbing of a film needs to be coordinated. The dubbing actors must coincide in the
takes in which their characters appear together. The dubbing actors are all paid the same
amount for each day they have to travel to the recording studio, regardless of the number
of takes recorded. It is not possible to record more than 6 takes per day. The goal is to
schedule the sessions per day so that the cost of the dubbing actors' services is as low as
possible

The cost function to minimize is based on:

- For each day, all the takes to be recorded are considered.
- If an actor is needed in more than one take on a specific day, they are counted only once, since the actor is paid per day.
- The actors needed for all the takes of each day are added up.
- Finally, these daily totals are added up to obtain the total cost.

Regarding the constraints, 5 working days with 6 takes each day are set.

```python
# Number of days and takes per day
days = 5
takes_per_day = 6
total_takes = matrix.shape[0]  # Total number of takes
```

#Analysis
- What is the complexity of the problem? Order of complexity and size of the solution space

We are considering the permutations of the 30 takes in groups of 6. The number of these permutations can be computed as the number of ways to arrange the 30 takes, divided by the number of ways to arrange the 6 takes within each group and the number of ways to arrange the 5 groups. According to the combinations formula, this number is:

$$
\large \frac{{30!}}{{(6!)^5 \cdot 5!}} = 11423951396577720
$$

This means that the brute force algorithm would have to generate and evaluate an extremely large number of candidate solutions, which would make the algorithm very slow and impractical.

```python
import math

# Compute the number of ways to split 30 takes into 5 groups of 6
num_ways = math.factorial(30) / (math.factorial(6)**5 * math.factorial(5))

# Print the number in decimal notation
print(format(num_ways, '.0f'))
```

Regarding the order of complexity, we can state that it is:

$$
\large O(n!)
$$

##Random Search

As is well known, random search does not produce good solutions; however, we will use it to understand the order of magnitude of the cost function we must minimize.

```python
# Number of times the algorithm will be run
num_runs = 10000

# Initialize the best schedule and the minimum cost
# These will be updated while the algorithm runs
best_schedule = None
min_cost = float('inf')

# Start measuring time
start_time = time.time()

# Run the algorithm num_runs times
# Each run generates a different schedule
for _ in range(num_runs):
    # Initialize the schedule and the set of scheduled takes
    # The schedule is a list of lists, where each list represents a day
    # scheduled_takes is a set that stores the takes that have already been scheduled
    schedule = []
    scheduled_takes = set()

    # Initialize the cost
    # The cost represents the number of conflicts in the schedule
    cost = 0

    # Generate the schedule
    # For each day, we select takes_per_day takes from the unscheduled takes
    for _ in range(days):
        day_schedule = []
        for _ in range(takes_per_day):
            # Select a random take from the unscheduled takes
            unscheduled_takes = [i for i in range(total_takes) if i not in scheduled_takes]
            take = random.choice(unscheduled_takes)

            # Add the take to the day's schedule and to the set of scheduled takes
            day_schedule.append(take)
            scheduled_takes.add(take)

        # Add the day's schedule to the overall schedule
        schedule.append(day_schedule)

        # Compute the cost of the day
        # The cost of the day is the number of conflicts on that day
        day_cost = np.any(matrix[np.array(day_schedule)], axis=0).sum()
        cost += day_cost

    # If the cost is lower than the minimum cost, update the best schedule and the minimum cost
    # This guarantees that we always keep the schedule with the minimum cost
    if cost < min_cost:
        best_schedule = schedule
        min_cost = cost

# Stop measuring time
end_time = time.time()

# Compute the execution time
execution_time = end_time - start_time

# Print the best schedule
# The best schedule is the schedule with the minimum cost
for i, day in enumerate(best_schedule):
    print(f'Day {i+1}: takes {tuple(take + 1 for take in day)}')

# Print the minimum cost
# The minimum cost is the cost of the best schedule
print(f'Minimum cost: {min_cost}')

# Print the execution time
print(f'Execution time: {execution_time} seconds')
```

##Tabu Search

- Initialization: The number of iterations (num_iterations) and the maximum size of the tabu list (tabu_size) are set.

- Generation of the initial schedule: A list of all the takes (all_takes) is created and randomly shuffled. Then, this list is split into several days, each day with a specific number of takes (takes_per_day). This is the initial schedule (schedule).

- Computation of the initial cost: The cost is the sum of the conflicts on each day of the schedule. A conflict occurs when two takes cannot happen on the same day. The initial cost is the sum of the conflicts in the initial schedule.

- Initialization of the best cost and the best schedule: At the beginning, the best cost is the initial cost and the best schedule is the initial schedule.

- Initialization of the tabu list: The tabu list is a list of moves that must not be made. At the beginning, the tabu list is empty.

- Tabu search: A tabu search is performed by iterating num_iterations times. In each iteration, the best move (swap of two takes) that is not in the tabu list and that reduces the cost is searched for. Then, this move is made, the cost and schedule are updated, and the move is added to the tabu list. If the tabu list exceeds its maximum size, the oldest move is removed. If the cost after making the move is lower than the best cost, the best cost and the best schedule are updated.

- At the end, the best schedule and the best cost are printed.

```python
num_iterations = 500  # Number of iterations the tabu search will perform
tabu_size = 50  # Maximum size of the tabu list

# Start measuring time
start_time = time.time()

# Generate a random initial schedule
all_takes = list(range(total_takes))  # Create a list of all the takes
random.shuffle(all_takes)  # Shuffle the list of takes
# Split the list of takes into days, each day with a specific number of takes
schedule = [all_takes[i:i+takes_per_day] for i in range(0, total_takes, takes_per_day)]

# Compute the initial cost
cost = sum(np.any(matrix[day], axis=0).sum() for day in schedule)

# Initialize the best cost and the best schedule
best_cost = cost  # The best cost found so far
best_schedule = schedule  # The best schedule found so far

# Initialize the tabu list
tabu_list = []  # List of moves that must not be made

# Tabu search: iterate num_iterations times
for _ in range(num_iterations):
    # Find the best move (swap of two takes) that is not in the tabu list
    best_move = None  # The best move found in this iteration
    best_move_cost = float('inf')  # The cost of the best move found in this iteration
    # Iterate over all pairs of takes in the schedule
    for i in range(days):
        for j in range(takes_per_day):
            for k in range(days):
                for l in range(takes_per_day):
                    # Swap two takes and compute the new cost
                    new_schedule = [list(day) for day in schedule]  # Create a copy of the current schedule
                    # Swap two takes in the new schedule
                    new_schedule[i][j], new_schedule[k][l] = new_schedule[k][l], new_schedule[i][j]
                    # Compute the cost of the new schedule
                    new_cost = sum(np.any(matrix[day], axis=0).sum() for day in new_schedule)

                    # If the new cost is lower and the move is not in the tabu list, update the best move
                    if new_cost < best_move_cost and (i, j, k, l) not in tabu_list:
                        best_move = (i, j, k, l)
                        best_move_cost = new_cost

    # Make the best move and update the cost and the schedule
    i, j, k, l = best_move  # Unpack the best move
    schedule[i][j], schedule[k][l] = schedule[k][l], schedule[i][j]  # Make the best move
    cost = best_move_cost  # Update the cost

    # Add the move to the tabu list
    tabu_list.append(best_move)  # Add the best move to the tabu list
    # If the tabu list exceeds its maximum size, remove the oldest move
    if len(tabu_list) > tabu_size:
        tabu_list.pop(0)

    # If the cost is lower than the best cost, update the best cost and the best schedule
    if cost < best_cost:
        best_cost = cost  # Update the best cost
        best_schedule = [list(day) for day in schedule]  # Update the best schedule

# Stop measuring time
end_time = time.time()

# Compute the execution time
execution_time = end_time - start_time

# Print the best schedule
for i, day in enumerate(best_schedule):
    print(f'Day {i+1}: takes {tuple(take + 1 for take in day)}')

# Print the best cost
print(f'Best cost: {best_cost}')

# Print the execution time
print(f'Execution time: {execution_time} seconds')
```

The tabu search algorithm is particularly useful for solving combinatorial optimization problems for several reasons:

- Exploration and exploitation: Tabu search balances the exploration of new solutions in the search space with the exploitation of the best solutions found so far. This can help avoid getting stuck in suboptimal local solutions.

- Memory: The tabu list acts as a short-term memory that stores recent moves. This prevents the algorithm from repeating the same moves and falling into infinite cycles, allowing a more diverse exploration of the search space.

- High-quality solutions: Although tabu search does not guarantee finding the global optimum, in practice it often produces high-quality solutions.

##Simulated Annealing

- Initialization: The algorithm starts with a random initial solution and a high “temperature”.

- Generation of new solutions: In each iteration, the algorithm generates a new solution from the current solution by making a small random change. In this case, the change is a swap of two takes in the schedule.

- Acceptance decision: The new solution is accepted or rejected based on its cost and the current temperature. If the new solution has a lower cost (it is better), it is accepted. If it has a higher cost (it is worse), it can be accepted with a probability that decreases with the cost difference and the temperature. This allows the algorithm to accept worse solutions at the beginning (when the temperature is high) to escape local optima, but it becomes increasingly selective as the temperature decreases.

- Cooling: After a certain number of iterations, the temperature is reduced according to a predefined cooling rate. This simulates the “cooling” in the annealing process.

- Termination: The algorithm ends after a predefined number of iterations or when the temperature falls below a certain threshold. The best solution found during the execution of the algorithm is returned as the result.

This code uses an exponential cooling function: $$ T_{\text{{new}}} = T_{\text{{current}}} \times \text{{cooling rate}} $$

Where:

- $T_{\text{{new}}}$ is the temperature after the current iteration.
- $ T_{\text{{current}}} $  is the temperature at the current iteration.

- $ \text{{cooling rate}} $ is a constant value lower than 1.

In each iteration of the algorithm, the temperature is multiplied by the cooling rate, which reduces the temperature exponentially as the algorithm progresses.

```python
# Simulated annealing parameters
num_iterations = 10000  # Number of iterations simulated annealing will perform
initial_temperature = 1000  # Initial temperature for simulated annealing
cooling_rate = 0.99  # Cooling rate for simulated annealing

# Generate a random initial schedule
all_takes = list(range(total_takes))  # Create a list of all the takes
random.shuffle(all_takes)  # Shuffle the list of takes
# Split the list of takes into days, each day with a specific number of takes
schedule = [all_takes[i:i+takes_per_day] for i in range(0, total_takes, takes_per_day)]

# Compute the initial cost
# The cost is the sum of the conflicts on each day of the schedule
cost = sum(np.any(matrix[day], axis=0).sum() for day in schedule)

# Initialize the best cost and the best schedule
# These will be updated while the algorithm runs
best_cost = cost  # The best cost found so far
best_schedule = schedule  # The best schedule found so far

# Start measuring time
start_time = time.time()

# Simulated annealing: iterate num_iterations times
temperature = initial_temperature  # Set the initial temperature
for _ in range(num_iterations):
    # Select two random takes to swap
    i, j = random.randint(0, days-1), random.randint(0, takes_per_day-1)
    k, l = random.randint(0, days-1), random.randint(0, takes_per_day-1)

    # Swap two takes and compute the new cost
    new_schedule = [list(day) for day in schedule]  # Create a copy of the current schedule
    new_schedule[i][j], new_schedule[k][l] = new_schedule[k][l], new_schedule[i][j]  # Swap two takes in the new schedule
    new_cost = sum(np.any(matrix[day], axis=0).sum() for day in new_schedule)  # Compute the cost of the new schedule

    # If the new cost is lower, or a worse solution is accepted with a certain probability, update the cost and the schedule
    # The probability of accepting a worse solution decreases with the temperature and the cost difference
    if new_cost < cost or random.random() < math.exp((cost - new_cost) / temperature):
        cost = new_cost
        schedule = new_schedule

    # If the cost is lower than the best cost, update the best cost and the best schedule
    if cost < best_cost:
        best_cost = cost
        best_schedule = [list(day) for day in schedule]

    # Cool down the temperature
    temperature *= cooling_rate

# Stop measuring time
end_time = time.time()

# Compute the execution time
execution_time = end_time - start_time

# Print the best schedule
for i, day in enumerate(best_schedule):
    print(f'Day {i+1}: takes {tuple(take + 1 for take in day)}')

# Print the best cost
print(f'Best cost: {best_cost}')

# Print the execution time
print(f'Execution time: {execution_time} seconds')
```

- Simulated annealing is a very flexible algorithm that can be adapted to many different types of optimization problems. It requires defining a “cost” function that can evaluate how good (or bad) a solution is, and a way to generate new solutions from an existing solution.

- Although simulated annealing does not guarantee finding the global optimum, it often finds very good solutions in a reasonable amount of time. This can be very useful when the total number of possible solutions is too large to examine each of them individually.
